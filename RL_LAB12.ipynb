```python
# REINFORCEMENT LEARNING - ASSIGNMENT 12
# Value Iteration in a Deterministic 2x2 Grid World

# 1. Define MDP Parameters

gamma = 0.9
normal_reward = -1
goal_reward = 100
theta = 0.000001

states = ["S0", "S1", "G"]
actions = ["Up", "Down", "Left", "Right"]

# 2. Define Transition Function

transitions = {
    "S0": {
        "Up": "S0",
        "Down": "S1",
        "Left": "S0",
        "Right": "S0"
    },
    "S1": {
        "Up": "S0",
        "Down": "S1",
        "Left": "S1",
        "Right": "G"
    }
}

# 3. Define Reward Function

def get_reward(next_state):
    if next_state == "G":
        return goal_reward
    return normal_reward

# 4. Initialize State Values

V = {
    "S0": 0,
    "S1": 0,
    "G": 0
}

# 5. Value Iteration Algorithm

iteration = 0

print("VALUE ITERATION")
print("--------------------------------")
print("Initial Values:", V)

while True:
    iteration += 1
    new_V = V.copy()
    for state in ["S0", "S1"]:
        q_values = []
        for action in actions:
            next_state = transitions[state][action]
            reward = get_reward(next_state)
            q_value = reward + gamma * V[next_state]
            q_values.append(q_value)
        new_V[state] = max(q_values)

    delta = max(abs(new_V[state] - V[state]) for state in ["S0", "S1"])
    V = new_V

    print(
        "Iteration", iteration,
        "| V(S0) =", round(V["S0"], 4),
        "| V(S1) =", round(V["S1"], 4),
        "| Delta =", round(delta, 6)
    )

    if delta < theta:
        break

# 6. Calculate Final Q Values

print("\nFINAL ACTION VALUES")
print("--------------------------------")

Q = {}
for state in ["S0", "S1"]:
    Q[state] = {}
    for action in actions:
        next_state = transitions[state][action]
        reward = get_reward(next_state)
        q_value = reward + gamma * V[next_state]
        Q[state][action] = q_value
        print(state, action, "->", next_state, "| Q =", round(q_value, 2))

# 7. Determine Optimal Policy

print("\nOPTIMAL POLICY")
print("--------------------------------")

policy = {}
for state in ["S0", "S1"]:
    best_action = max(Q[state], key=Q[state].get)
    policy[state] = best_action
    print(state, "->", best_action, "| Value =", round(V[state], 2))

# 8. Display Final Results

print("\nFINAL OPTIMAL STATE VALUES")
print("--------------------------------")
for state in states:
    print(state, "=", round(V[state], 2))

print("\nOPTIMAL PATH")
print("--------------------------------")
print("S0 -> S1 -> G")

print("\nValue Iteration Completed Successfully!")
```
