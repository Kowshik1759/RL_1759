```python
# Reinforcement Learning - Assignment 11
# Policy Iteration in a Deterministic 2x2 Grid World

gamma = 0.9
goal_reward = 100
step_reward = -1

states = ["S0", "S1"]
actions = ["Up", "Down", "Left", "Right"]

# Initial policy
policy = {
    "S0": "Up",
    "S1": "Up"
}

# Deterministic transition function
def transition(state, action):
    if state == "S0":
        if action == "Down":
            return "S1", step_reward
        else:
            return "S0", step_reward
    elif state == "S1":
        if action == "Right":
            return "G", goal_reward
        elif action == "Up":
            return "S0", step_reward
        else:
            return "S1", step_reward

# Policy Evaluation
def policy_evaluation(policy):
    V = {"S0": 0, "S1": 0, "G": 0}
    for iteration in range(1000):
        old_values = V.copy()
        for state in states:
            next_state, reward = transition(state, policy[state])
            V[state] = reward + gamma * old_values[next_state]
        if all(abs(V[s] - old_values[s]) < 1e-10 for s in states):
            break
    return V

# Calculate action value Q(s,a)
def calculate_q(state, action, V):
    next_state, reward = transition(state, action)
    return reward + gamma * V[next_state]

# Policy Iteration
iteration = 1
while True:
    print("\n========== POLICY ITERATION", iteration, "==========")
    # Step 1: Policy Evaluation
    V = policy_evaluation(policy)
    print("\nPolicy:", policy)
    print("State Values:", V)

    # Step 2: Calculate Q values
    print("\nAction Values:")
    for state in states:
        for action in actions:
            q = calculate_q(state, action, V)
            print(state, action, "=", round(q, 2))

    # Step 3: Policy Improvement
    new_policy = {}
    # Tie-breaking order: prefer Down at S0
    priority = ["Down", "Right", "Up", "Left"]
    for state in states:
        q_values = {
            action: calculate_q(state, action, V)
            for action in actions
        }
        best_action = max(
            priority,
            key=lambda action: q_values[action]
        )
        new_policy[state] = best_action
    print("\nImproved Policy:", new_policy)

    # Step 4: Check policy stability
    if new_policy == policy:
        print("\nPolicy is stable!")
        break
    policy = new_policy
    iteration += 1

# Final Results
print("\n========== FINAL RESULTS ==========")
print("Optimal Policy:", policy)
print("Optimal State Values:", V)
print("\nOptimal Path: S0 -> S1 -> G")

# Bellman Optimality Verification
print("\n========== BELLMAN VERIFICATION ==========")
for state in states:
    q_values = [
        calculate_q(state, action, V)
        for action in actions
    ]
    print(state, "Maximum Q Value =", round(max(q_values), 2))
print("Goal State Value = 0")
print("\nPolicy Iteration Completed Successfully!")
```
