```python
# REINFORCEMENT LEARNING - ASSIGNMENT 7
# Assignment 1: Discounted and Undiscounted Returns
# Assignment 2: MDP Representation of 2x3 Grid World

# ==========================================
# ASSIGNMENT 1: RETURN CALCULATION
# ==========================================

print("===== ASSIGNMENT 1: RETURN CALCULATION =====")

# Fixed number of steps and rewards
rewards = [10, 20, 30, 40]
gamma = 0.9
n = len(rewards)

# Calculate undiscounted return
undiscounted_return = sum(rewards)

# Calculate discounted return
discounted_return = sum(
    (gamma ** t) * rewards[t] for t in range(n)
)

print("Number of Steps:", n)
print("Rewards:", rewards)
print("Discount Factor:", gamma)
print("Undiscounted Return:", undiscounted_return)
print("Discounted Return:", round(discounted_return, 2))


# ==========================================
# ASSIGNMENT 2: MDP GRID WORLD
# ==========================================

print("\n===== ASSIGNMENT 2: MDP GRID WORLD =====")

states = ["S0", "S1", "S3", "S4", "G"]
actions = ["Up", "Down", "Left", "Right"]

gamma = 0.9
normal_reward = -1
goal_reward = 10
obstacle_reward = -5

# Transition table
transitions = {
    "S0": {
        "Up": "S0",
        "Down": "S0",
        "Left": "S0",
        "Right": "S1"
    },
    "S1": {
        "Up": "S1",
        "Down": "S3",
        "Left": "S0",
        "Right": "G"
    },
    "S3": {
        "Up": "S1",
        "Down": "S3",
        "Left": "S3",
        "Right": "S4"
    },
    "S4": {
        "Up": "G",
        "Down": "S4",
        "Left": "S3",
        "Right": "S4"
    }
}

# Transition function
def transition(state, action):
    if state == "G":
        return "G"
    return transitions[state][action]

# Reward function
def get_reward(next_state):
    if next_state == "G":
        return goal_reward
    return normal_reward

# Display MDP details
print("State Space:", states)
print("Action Space:", actions)
print("Discount Factor:", gamma)

print("\nTransition Table:")

for state in transitions:
    for action in actions:
        next_state = transition(state, action)
        reward = get_reward(next_state)
        print(state, "+", action, "->", next_state, "| Reward:", reward)

# Fixed example transition
current_state = "S0"
action = "Right"

next_state = transition(current_state, action)
reward = get_reward(next_state)

print("\n===== TRANSITION RESULT =====")
print("Current State:", current_state)
print("Selected Action:", action)
print("Next State:", next_state)
print("Reward:", reward)

print("\n===== PROGRAM COMPLETED =====")
```
